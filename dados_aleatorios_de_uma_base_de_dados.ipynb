```python
import pandas as pd

# Lê o arquivo CSV (substitua 'seu_arquivo.csv' pelo nome do seu arquivo)
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Concessionaria/DadosVendasCarros.CSV',encoding='utf-8', sep=';')

#display(df)

# Agrupa os dados por 'marca' e 'modelo' e pega amostras aleatórias
amostra = df.groupby(['MARCA', 'REGIÃO','MODELO']).apply(lambda x: x.sample(n=1)).reset_index(drop=True).sample(n=10)
# Exibe os 10 registros
cols=['MARCA', 'MODELO', 'PAÍS']
display(amostra[cols])
```
